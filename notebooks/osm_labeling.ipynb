```python
import osmnx as ox

# Use a faster Overpass endpoint
ox.settings.overpass_url = "https://overpass.kumi.systems/api"
ox.settings.timeout = 30  # shorter timeout so it fails faster

# Coordinates near a known industrial area in Berlin
lat, lon = 52.515, 13.389

print("Querying at:", lat, lon)

try:
    industrial = ox.features_from_point(
        (lat, lon),
        tags={"landuse": "industrial"},
        dist=200  # smaller radius to reduce load
    )
    print("Result shape:", industrial.shape)
    print(industrial.head())
except Exception as e:
    print("Query failed:", e)
```
