```python
import pandas as pd

INPUT_FILE = "archive.csv"
OUTPUT_FILE = "india_archive.csv"

chunksize = 100000

first_chunk = True

for chunk in pd.read_csv(INPUT_FILE, chunksize=chunksize):

    india = chunk[
        (chunk["latitude"] >= 6) &
        (chunk["latitude"] <= 38) &
        (chunk["longitude"] >= 68) &
        (chunk["longitude"] <= 98)
    ]

    india.to_csv(
        OUTPUT_FILE,
        mode="w" if first_chunk else "a",
        header=first_chunk,
        index=False
    )

    first_chunk = False

print("Done!")
```
